**Genre Trends:**

Analyze trends in the popularity of different genres over time.

```python
# Split genres and explode
df_genre = df.assign(genre=df['genres'].str.split(', ')).explode('genre')

# Count total occurrences of each genre
top5_genres = df_genre['genre'].value_counts().head(5).index

# Filter only top 5 genres
df_top5 = df_genre[df_genre['genre'].isin(top5_genres)]

# Group by year and genre
genre_trends = df_top5.groupby(['release_year', 'genre']).size().reset_index(name='count')

# --- Visualization ---

plt.figure(figsize=(12,6))

for genre in top5_genres:
    temp = genre_trends[genre_trends['genre'] == genre]
    plt.plot(temp['release_year'], temp['count'], label=genre)

plt.title("Popularity Trends of Top 5 Genres Over Time")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.legend()
plt.tight_layout()
plt.show()
```

1. Significant Increase in Genre Diversity Over Time

The number of titles across all major genres shows a clear upward trend, especially after 2015. Early years had relatively fewer releases, whereas later years reflect a broader and more diverse content catalog. This indicates Netflix’s strategic expansion and higher production volume in recent years.

2. Rapid Growth in International Content

International genres demonstrate a notable rise, particularly after 2018. This trend highlights Netflix’s global expansion strategy and increasing investment in region-specific content to cater to non-English-speaking markets.

3. Steady Rise in Documentary Content

Documentary titles exhibit consistent growth across the years. The steady upward trend suggests increased audience interest in factual, reality-based storytelling and Netflix’s focus on producing high-quality documentary films and series.

4. Dominance of Drama-Based Genres

Drama remains one of the most consistently popular and frequently released genres throughout the observed period. Despite fluctuations in other genres, drama continues to hold a central place in Netflix’s content lineup, reflecting stable viewer demand for narrative-driven stories.

5. Increasing Popularity of Crime-Related Genres

Crime genres show a noticeable surge, particularly after 2016. This rise aligns with the growing popularity of crime thrillers and investigative series on streaming platforms, influencing Netflix to expand content production in this category.

6. Sharp Increase Around 2020

Most genres exhibit a pronounced spike in the number of releases around 2020. This trend corresponds to the COVID-19 pandemic period, during which streaming consumption increased significantly, prompting Netflix to accelerate the release of new content across multiple genres.

**Content Evolution Over Time:**

STEP-15: EXPLORE HOW THE CHARACTERISTICS OF
CONTENT (e.g., DURATION, RATINGS) HAVE EVOLVED
OVER THE YEARS.

```python
# Filter movies with numeric duration (in minutes)
duration_over_time = df[df['duration'].str.contains('min',
na=False)].copy()
duration_over_time['duration_mins'] =duration_over_time['duration'].str.extract('(\d+)').astype(float)
# Group by release year and calculate average duration
avg_duration_by_year = duration_over_time.groupby('release_year')['duration_mins'].mean()
plt.figure(figsize=(12, 6))
avg_duration_by_year.plot(marker='o', color='teal')
plt.title('Average Movie Duration Over the Years',fontsize=20)
plt.xlabel('Release Year',fontsize=16)
plt.ylabel('Average Duration (minutes)',fontsize=16)
plt.grid(True)
plt.tight_layout()
plt.show()
```

✅Insights:

• A declining trend in average movie duration is observed in recent years, suggesting a
shift toward shorter, more consumable content.

• Older movies (pre-2010) had longer average runtimes, indicating a past preference for
extended storytelling.

• This shift may reflect changing viewer behavior, with modern audiences favoring concise,
fast-paced formats.

```python
# --- 2. Rating Trend Over Years ---
top_ratings = df['rating'].value_counts().head(6).index
df_r = df[df['rating'].isin(top_ratings)]

pivot_rating = df_r.pivot_table(index='release_year',columns='rating',
    values='show_id',aggfunc='count',fill_value=0)
plt.figure(figsize=(12,8))
pivot_rating.plot()
plt.title("Rating Distribution Over Years")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.tight_layout()
plt.show()
```

**1. Evolution of Movie Duration Over Time**

The average duration of movies shows fluctuations across the years, indicating no strong linear trend toward longer or shorter films.

In certain periods, average duration increases slightly, suggesting production of more full-length movies.

Some years show a drop in average movie length, possibly due to an increase in shorter films, documentaries, or stand-up specials.

Overall, movie duration does not show a consistent upward or downward trend, implying that content length has remained relatively stable over time.

**2. Evolution of Ratings Distribution Over Time**

Over the years, certain ratings (such as TV-MA and TV-14) show an increase, reflecting a rise in mature and young-adult oriented content.

Ratings like G, PG, and TV-Y appear less frequently compared to mature categories, indicating a shift toward content aimed at older audiences.

The growing number of TV-MA titles suggests that Netflix has increasingly focused on producing or acquiring more adult-oriented content.

The spread of ratings over the years indicates greater content variety, but the trend leans toward more intense, mature, or complex themes.

**User Preferences:**

Investigate whether certain genres or types of content are more popular among users.

```python
rating_map = {
    'G': 1, 'TV-Y': 1, 'TV-G': 1,
    'PG': 2, 'TV-PG': 2, 'TV-Y7': 2,
    'PG-13': 3, 'TV-14': 3,
    'R': 4, 'TV-MA': 4,
    'NC-17': 5
}

df['rating_numeric'] = df['rating'].map(rating_map)

# ---------------------------------------
# Genre popularity (average rating)
# ---------------------------------------
df['genre_list'] = df['genres'].str.split(',')
genre_exploded = df.explode('genre_list')
genre_exploded['genre_list'] = genre_exploded['genre_list'].str.strip()

avg_genre_rating = genre_exploded.groupby('genre_list')['rating_numeric'].mean().sort_values(ascending=False)

plt.figure(figsize=(12,6))
sns.barplot(
    x=avg_genre_rating.values,
    y=avg_genre_rating.index,
    palette="mako"
)
plt.title("Average Rating by Genre (Popularity Proxy)")
plt.xlabel("Average Rating (Numeric)")
plt.ylabel("Genre")
plt.tight_layout()
plt.show()

# ---------------------------------------
# Type popularity (Movie vs TV Show)
# ---------------------------------------
avg_type_rating = df.groupby('type')['rating_numeric'].mean()

plt.figure(figsize=(8,5))
sns.barplot(
    x=avg_type_rating.index,
    y=avg_type_rating.values,
    palette="mako"
)
plt.title("Average Rating by Content Type")
plt.xlabel("Content Type")
plt.ylabel("Average Rating (Numeric)")
plt.tight_layout()
plt.show()
```

1. Genre Popularity Insights

Genres with higher average numeric ratings—such as Documentaries, Dramas, and certain International genres—appear to be the most preferred among users.

These higher ratings indicate that viewers tend to perceive such content as more mature, impactful, or appealing.

Genres with lower average ratings often include lighter or youth-oriented content, suggesting they cater to a more general or younger audience with less intensity in themes.

The variation in average ratings across genres indicates a clear difference in perceived quality or user preference, with serious or informational genres ranking higher.

2. Popularity of Content Types (Movies vs TV Shows)

The average ratings of Movies vs TV Shows may show a slight difference, with one format receiving higher ratings than the other.

If movies have a higher average score, it suggests users favor the structure and pacing of films.

If TV shows score higher, it implies that users prefer long-form storytelling or that Netflix invests more in high-quality series.

Overall, the difference in average rating between the two content types reveals user leanings toward one format over the other in terms of perceived quality.
